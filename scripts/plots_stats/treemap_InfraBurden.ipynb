```python
df_0 = pd.read_csv('df_ssp2_clean.csv', index_col= 0)
```

```python
volm3_bins = [0, 250, 500, 1000, 5000, 400000]
volm3_labels = ['0-250 m³', '250-500 m³', '500-1000 m³', '1000-5000 m³', '5000+ m³']
len_m_bins = [0, 5, 20, 50, 100, 15000]
len_m_labels = ['0-5 m', '5-20 m', '20-50 m', '50-100 m', '100+ m']
df_0['range_volm3_2020'] = pd.cut(df_0['volume_m3_perCap_2020_ssp2'], bins = volm3_bins, labels = volm3_labels, right=True)
df_0['range_volm3_2050'] = pd.cut(df_0['volume_m3_perCap_2050_ssp2'], bins = volm3_bins, labels = volm3_labels, right=True)
df_0['range_volm3_2100'] = pd.cut(df_0['volume_m3_perCap_2100_ssp2'], bins = volm3_bins, labels = volm3_labels, right=True)
df_0['range_len_m_2020'] = pd.cut(df_0['length_m_perCap_2020_ssp2'], bins = volm3_bins, labels = volm3_labels, right=True)
df_0['range_len_m_2050'] = pd.cut(df_0['length_m_perCap_2050_ssp2'], bins = volm3_bins, labels = volm3_labels, right=True)
df_0['range_len_m_2100'] = pd.cut(df_0['length_m_perCap_2100_ssp2'], bins = volm3_bins, labels = volm3_labels, right=True)
```

```python
from plotly.graph_objs import Line
# import plotly.io as pio

# pio.get_chrome()

def plot_treemap(df, groupby_cols, color_column, filename):
  df = df[df[groupby_cols[0]] != 'noChange']
  df_infra_year = df.groupby(groupby_cols,observed=False).size().reset_index(name='count')
  # --- Create Treemap ---
  fig = px.treemap(
      df_infra_year ,
      path=groupby_cols,
      values="count",
      color=color_column,
      title="burden by city type and per capita range",
      color_discrete_map={'urban':'red', 'suburban':'lightseagreen', 'periurban':'dimgrey', 'rural':'darkkhaki',
                          'increasingBurden':'azure'},
      )
  # fig.data[0].textinfo = 'label+text+value'
  # Set width and height
  fig.update_layout(
      width= 600,   # adjust width as needed
      height=600   # adjust height as needed
  )
  # Update the marker line properties
  fig.update_traces(
      marker_line_width=0.1,
  )
  fig.update_layout(
    uniformtext=dict(
        minsize=16,  # Set the desired minimum font size
        # mode='show'  # 'show' will display text even if it overflows, 'hide' will hide it
    )
  )
  # print(fig.data[0])
  scale_factor = 4  # Adjust this value based on your requirements
  fig.update_traces(sort=False)
  fig.update_layout(autosize=False, width=600, height=600) # Adjust width and height as needed
  fig.write_image(filename, scale=scale_factor)
  fig.show()
```

```python
df = df_0.copy()

group_cols = ['RBUV_Burden_2050','citytype_at_2050','range_volm3_2050']
color_col = 'citytype_at_2050'
plot_treemap(df, group_cols, color_col, 'RBUV_2050.png')

group_cols = ['RBUV_Burden_2100','citytype_at_2100','range_volm3_2100']
color_col = 'citytype_at_2100'
plot_treemap(df, group_cols, color_col, 'RBUV_2100.png')

group_cols = ['RL_Burden_2050','citytype_at_2050','range_len_m_2050']
color_col = 'citytype_at_2050'
plot_treemap(df, group_cols, color_col, 'RL_2050.png')

group_cols = ['RL_Burden_2100','citytype_at_2100','range_len_m_2100']
color_col = 'citytype_at_2100'
plot_treemap(df, group_cols, color_col, 'RL_2100.png')
```
